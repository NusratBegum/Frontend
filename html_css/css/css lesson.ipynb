# CSS: Classes

    1. classes: 
- Use classes when you want to create a consistent style pattern that you will repeat across your website (like buttons, card layouts, or text colors)
- 


```python
#html
#<!-- You can use the same class on different elements -->
<button class="btn highlight">Submit</button>
<p class="highlight">Important note here!</p>

#Note: You can apply multiple classes to a single element by separating them with a space (e.g., class="btn highlight"

#css
#/* Target a class using a dot (.) */
.highlight {
  background-color: yellow;
  font-weight: bold;
}


```

# CSS ID (#):
- Use an ID when an element is completely unique to the page and will never be duplicated (like a header navbar, a side sidebar, or a main contact form

```python
#html example:
<!-- This ID must be unique to this specific page -->
<div id="main-header">Welcome to my site</div>

#css example:
#/* Target an ID using a hash (#) */
#main-header {
  background-color: #333;
  color: white;
  padding: 20px;
}
```
